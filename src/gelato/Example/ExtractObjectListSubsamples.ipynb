When the gelato fit production is interrupted one need a subsample of objects files (fits) remaining to be processed.


```python
import pandas as pd
from astropy.io import fits
from astropy.table import Table
import re
```

## Config

```python
version_table = "v3"
```

```python
obj_filename_fits =  f"object_filelist_{version_table}.fits" 
short_obj_filename_fits =  f"object_filelist_{version_table}_short.fits" 
```

```python
flag_save_filelist = False
```

## Read full object list

```python
t = Table.read(obj_filename_fits )
```

```python
df = t.to_pandas()
```

```python
df
```

### For v3 version
- object_filelist_v3.fits
- restart after  specgelato_SPEC611.fits

```python
if version_table == "v3":
    tag = b".*611.*"
```

### For v4 version
- object_filelist_v4.fits
- restart after specgelato_SPEC381.fits

```python
if version_table == "v4":
    tag = b".*381.*"
```

## Seach the index

```python
for idx,fn in  enumerate(df.Path.values):
    if len(re.findall(tag,fn))>0:
        break
print(idx,df.iloc[idx])
```

## Drop lines

```python
df_out = df.drop(index=range(0,idx))
```

```python
df_out
```

```python
t_out = Table.from_pandas(df_out)
```

```python
t_out
```

```python
if flag_save_filelist:
    t_out.write(short_obj_filename_fits,format="fits",overwrite=True)
```
